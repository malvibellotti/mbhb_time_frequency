```python
import os
import glob
import json
import numpy as np
import matplotlib.pyplot as plt
import corner
from scipy.stats import gaussian_kde
from bbhx.utils.constants import PC_SI

# --- 1. COPY THE PARSER FUNCTIONS HERE ---
def extract_alberto_json(filepath):
    with open(filepath, 'r') as f:
        data = json.load(f)
    p = data['source_params']
    mc, q_json = float(p['Mchirp']), float(p['q'])
    m_A = mc * (1.0 + q_json)**0.2 / (q_json**0.6)       
    m_B = m_A * q_json
    m1, m2 = max(m_A, m_B), min(m_A, m_B)
    dist = float(p['dist']) * 1e6 * PC_SI
    inc = float(p['inc'])
    return [m1, m2/m1, dist, inc] # We only need these 4 for the KDE test

def load_catalog_for_test(repo_path, pop_name="Q3d"):
    json_dir = os.path.join(repo_path, pop_name, "json")
    json_files = glob.glob(os.path.join(json_dir, "*.json"))
    params =[]
    for jf in json_files:
        try:
            params.append(extract_alberto_json(jf))
        except: pass
    return np.array(params)

# --- 2. LOAD DATA ---
repo_path = "/sps/lisaf/mbellotti/github/standard-sirens-MBHBs" 
print("Loading catalog...")

pop_name = "Q3nd"  # Change to "Q3d", "Q3nd", or "Pop3" as needed
real_params = load_catalog_for_test(repo_path, pop_name)

# --- 3. FIT KDE & SAMPLE ---
print("Fitting KDE...")
log_m1 = np.log10(real_params[:, 0])
q = real_params[:, 1]
log_dist = np.log10(real_params[:, 2])
inc = real_params[:, 3]

training_data = np.vstack([log_m1, q, log_dist, inc])
kde = gaussian_kde(training_data)

print("Generating synthetic samples...")
synthetic_data = kde.resample(5000)
synthetic_data[1, :] = np.clip(synthetic_data[1, :], 0.01, 1.0) # Clip q
synthetic_data[3, :] = np.clip(synthetic_data[3, :], 0.0, np.pi) # Clip inc

# --- 4. PLOT AND COMPARE ---
labels = [r"$\log_{10}(m_1)$", r"$q$", r"$\log_{10}(D_L)$", r"$\iota$"]

# Plot the synthetic KDE samples in Orange
fig = corner.corner(synthetic_data.T, color='darkorange', labels=labels, 
                    hist_kwargs={'density': True}, plot_datapoints=False, fill_contours=True)

# Overplot Alberto's true catalog in Blue
corner.corner(training_data.T, fig=fig, color='royalblue', 
              hist_kwargs={'density': True}, plot_datapoints=False, fill_contours=True)

fig.suptitle(f"blue: {pop_name} | orange: sampled prior", fontsize=16)
plt.show()
```

```python
import os
import glob
import json
import h5py
import numpy as np
import matplotlib.pyplot as plt
import corner
from scipy.stats import gaussian_kde
from bbhx.utils.constants import PC_SI, YRSID_SI

repo_path = "/sps/lisaf/mbellotti/github/standard-sirens-MBHBs"
pop_name = "Q3d"  # Change to "Q3nd" or "Pop3" to see the others!

# ======================================================================
# 1. CHECK THE MODES IN ALBERTO'S CATALOG
# ======================================================================
h5_path = os.path.join(repo_path, "data_for_Malvina.h5")
print(f"--- Checking Modes for {pop_name} ---")

if os.path.exists(h5_path):
    with h5py.File(h5_path, 'r') as f:
        if pop_name in f:
            pop_group = f[pop_name]
            mode_counts = {1: 0, 2: 0, 8: 0, "Unknown": 0}
            
            for strid in pop_group.keys():
                ev = pop_group[strid]
                if ev['1mode_5p'][()] == 1: mode_counts[1] += 1
                elif ev['2modes_5p'][()] == 1: mode_counts[2] += 1
                elif ev['8modes_5p'][()] == 1: mode_counts[8] += 1
                else: mode_counts["Unknown"] += 1
                
            print(f"Total Events with Posteriors: {len(pop_group.keys())}")
            print(f"  -> 1-Mode Events (Perfect Localization) : {mode_counts[1]}")
            print(f"  -> 2-Mode Events (Bi-modal Degeneracy)  : {mode_counts[2]}")
            print(f"  -> 8-Mode Events (Full Sky Degeneracy)  : {mode_counts[8]}")
            print(f"  -> Unclassified Events                  : {mode_counts['Unknown']}\n")

# ======================================================================
# 2. EXTRACT ALL 8 PARAMETERS
# ======================================================================
def extract_all_params(filepath):
    with open(filepath, 'r') as f:
        data = json.load(f)
    p = data.get('source_params', data.get('params', data.get('injection', {})))
    
    def get_val(keys):
        for k in keys:
            if k in p: return float(p[k])
        return 0.0

    mc, q_json = get_val(['Mchirp', 'mc']), get_val(['q', 'mass_ratio'])
    m_A = mc * (1.0 + q_json)**0.2 / (q_json**0.6)       
    m1, m2 = max(m_A, m_A * q_json), min(m_A, m_A * q_json)
    q = m2 / m1 if m1 > 0 else 1.0
    
    dist = get_val(['dist', 'DL_Mpc', 'dL']) * 1e6 * PC_SI
    
    # CRITICAL FIX: Use Tc (Merger Time), NOT Deltat! Convert to years.
    t_ref = get_val(['Tc', 'tC', 'tc', 't_ref', 'time']) / YRSID_SI
    
    # Extract and WRAP the angles properly
    inc = get_val(['inc', 'iota'])
    lam = np.mod(get_val(['lam', 'lambda', 'phS']), 2 * np.pi)
    beta = np.pi/2.0 - p['thS'] if 'thS' in p else get_val(['beta', 'dec'])
    psi = np.mod(get_val(['psi', 'pol']), np.pi)
    
    return [np.log10(m1), q, np.log10(dist), t_ref, beta, lam, inc, psi]

print(f"--- Loading full catalog JSONs for {pop_name} ---")
json_dir = os.path.join(repo_path, pop_name, "json")
json_files = glob.glob(os.path.join(json_dir, "*.json"))

catalog_data = []
for jf in json_files:
    try:
        catalog_data.append(extract_all_params(jf))
    except: pass
catalog_data = np.array(catalog_data)
print(f"Extracted {len(catalog_data)} events.")

# ======================================================================
# 3. FIT KDE & GENERATE SYNTHETIC PRIOR
# ======================================================================
print("Fitting 8-Dimensional KDE...")
kde = gaussian_kde(catalog_data.T)

print("Generating 5000 synthetic samples...")
synthetic_data = kde.resample(5000).T

# Clip the synthetic samples so they don't break physical boundaries
synthetic_data[:, 1] = np.clip(synthetic_data[:, 1], 0.01, 1.0)           # q
synthetic_data[:, 4] = np.clip(synthetic_data[:, 4], -np.pi/2, np.pi/2)   # beta
synthetic_data[:, 5] = np.clip(synthetic_data[:, 5], 0.0, 2*np.pi)        # lam
synthetic_data[:, 6] = np.clip(synthetic_data[:, 6], 0.0, np.pi)          # inc
synthetic_data[:, 7] = np.clip(synthetic_data[:, 7], 0.0, np.pi)          # psi

# ======================================================================
# 4. PLOT AND COMPARE
# ======================================================================
labels = [r"$\log_{10}(m_1)$", r"$q$", r"$\log_{10}(D_L)$", r"$t_{ref}$ [Yrs]", r"$\beta$", r"$\lambda$", r"$\iota$", r"$\psi$"]

print("Plotting Massive Corner Plot...")
# Plot the synthetic KDE samples in Orange
fig = corner.corner(synthetic_data, color='darkorange', labels=labels, 
                    hist_kwargs={'density': True}, plot_datapoints=False, fill_contours=True, smooth=0.8)

# Overplot Alberto's true catalog in Blue
corner.corner(catalog_data, fig=fig, color='royalblue', 
              hist_kwargs={'density': True}, plot_datapoints=False, fill_contours=True, smooth=0.8)

# Custom Legend
from matplotlib.lines import Line2D
custom_lines = [
    Line2D([0], [0], color='royalblue', lw=4, alpha=0.5, label=f"Alberto's True {pop_name} Catalog"),
    Line2D([0], [0], color='darkorange', lw=4, alpha=0.5, label='KDE Sampled ML Prior')
]
fig.legend(handles=custom_lines, loc='upper right', fontsize=16, bbox_to_anchor=(0.95, 0.95))

fig.suptitle(f"Astrophysical Prior Distribution Check ({pop_name})", fontsize=22, y=1.02)
plt.show()
```

```python
import os
import glob
import json
import h5py
import numpy as np
import matplotlib.pyplot as plt
import corner
from scipy.stats import gaussian_kde
from bbhx.utils.constants import PC_SI, YRSID_SI

repo_path = "/sps/lisaf/mbellotti/github/standard-sirens-MBHBs"
pop_name = "Q3d"  # Change to "Q3nd" or "Pop3" to see the others!

# ======================================================================
# 1. CHECK THE MODES IN ALBERTO'S CATALOG
# ======================================================================
h5_path = os.path.join(repo_path, "data_for_Malvina.h5")
print(f"--- Checking Modes for {pop_name} ---")

if os.path.exists(h5_path):
    with h5py.File(h5_path, 'r') as f:
        if pop_name in f:
            pop_group = f[pop_name]
            mode_counts = {1: 0, 2: 0, 8: 0, "Unknown": 0}
            
            for strid in pop_group.keys():
                ev = pop_group[strid]
                if ev['1mode_5p'][()] == 1: mode_counts[1] += 1
                elif ev['2modes_5p'][()] == 1: mode_counts[2] += 1
                elif ev['8modes_5p'][()] == 1: mode_counts[8] += 1
                else: mode_counts["Unknown"] += 1
                
            print(f"Total Events with Posteriors: {len(pop_group.keys())}")
            print(f"  -> 1-Mode Events (Perfect Localization) : {mode_counts[1]}")
            print(f"  -> 2-Mode Events (Bi-modal Degeneracy)  : {mode_counts[2]}")
            print(f"  -> 8-Mode Events (Full Sky Degeneracy)  : {mode_counts[8]}")
            print(f"  -> Unclassified Events                  : {mode_counts['Unknown']}\n")

# ======================================================================
# 2. EXTRACT ALL 8 PARAMETERS
# ======================================================================
def extract_all_params(filepath):
    with open(filepath, 'r') as f:
        data = json.load(f)
    p = data.get('source_params', data.get('params', data.get('injection', {})))
    
    def get_val(keys):
        for k in keys:
            if k in p: return float(p[k])
        return 0.0

    mc, q_json = get_val(['Mchirp', 'mc', 'mchirp']), get_val(['q', 'mass_ratio'])
    m_A = mc * (1.0 + q_json)**0.2 / (q_json**0.6)       
    m1, m2 = max(m_A, m_A * q_json), min(m_A, m_A * q_json)
    q = m2 / m1 if m1 > 0 else 1.0
    
    dist = get_val(['dist', 'DL_Mpc', 'dL', 'D_L']) * 1e6 * PC_SI
    
    # Check all possible time keys!
    t_ref = get_val(['Tc', 'tC', 'tc', 't_ref', 'time', 'Deltat']) / YRSID_SI
    
    inc = get_val(['inc', 'iota'])
    lam = np.mod(get_val(['lam', 'lambda', 'phS']), 2 * np.pi)
    beta = np.pi/2.0 - p['thS'] if 'thS' in p else get_val(['beta', 'dec'])
    psi = np.mod(get_val(['psi', 'pol']), np.pi)
    
    return [np.log10(m1), q, np.log10(dist), t_ref, beta, lam, inc, psi]

print(f"--- Loading full catalog JSONs for {pop_name} ---")
json_dir = os.path.join(repo_path, pop_name, "json")
json_files = glob.glob(os.path.join(json_dir, "*.json"))

catalog_data = []
for jf in json_files:
    try:
        catalog_data.append(extract_all_params(jf))
    except: pass
catalog_data = np.array(catalog_data)

# CRITICAL FIX: Add a tiny amount of jitter to prevent the LinAlgError 
# if a parameter happens to be identical across all events!
jitter = np.random.randn(*catalog_data.shape) * 1e-6
catalog_data += jitter

print(f"Extracted {len(catalog_data)} events.")

# ======================================================================
# 3. FIT KDE & GENERATE SYNTHETIC PRIOR
# ======================================================================
print("Fitting 8-Dimensional KDE...")
kde = gaussian_kde(catalog_data.T)

print("Generating 5000 synthetic samples...")
synthetic_data = kde.resample(5000).T

# Clip the synthetic samples so they don't break physical boundaries
synthetic_data[:, 1] = np.clip(synthetic_data[:, 1], 0.01, 1.0)           # q
synthetic_data[:, 4] = np.clip(synthetic_data[:, 4], -np.pi/2, np.pi/2)   # beta
synthetic_data[:, 5] = np.clip(synthetic_data[:, 5], 0.0, 2*np.pi)        # lam
synthetic_data[:, 6] = np.clip(synthetic_data[:, 6], 0.0, np.pi)          # inc
synthetic_data[:, 7] = np.clip(synthetic_data[:, 7], 0.0, np.pi)          # psi

# ======================================================================
# 4. PLOT AND COMPARE
# ======================================================================
labels = [r"$\log_{10}(m_1)$", r"$q$", r"$\log_{10}(D_L)$", r"$t_{ref}$ [Yrs]", r"$\beta$", r"$\lambda$", r"$\iota$", r"$\psi$"]

print("Plotting Massive Corner Plot...")
fig = corner.corner(synthetic_data, color='darkorange', labels=labels, 
                    hist_kwargs={'density': True}, plot_datapoints=False, fill_contours=True, smooth=0.8)

corner.corner(catalog_data, fig=fig, color='royalblue', 
              hist_kwargs={'density': True}, plot_datapoints=False, fill_contours=True, smooth=0.8)

from matplotlib.lines import Line2D
custom_lines = [
    Line2D([0], [0], color='royalblue', lw=4, alpha=0.5, label=f"Alberto's True {pop_name} Catalog"),
    Line2D([0], [0], color='darkorange', lw=4, alpha=0.5, label='KDE Sampled ML Prior')
]
fig.legend(handles=custom_lines, loc='upper right', fontsize=16, bbox_to_anchor=(0.95, 0.95))

fig.suptitle(f"Astrophysical Prior Distribution Check ({pop_name})", fontsize=22, y=1.02)
plt.show()
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# Point this to the file you just generated!
h5_path = "/sps/lisaf/mbellotti/github/phd_disentangling/scripts/dataset.h5" 

with h5py.File(h5_path, 'r') as f:
    data = f['data'][0] # Load first event
    params = f['parameters'][0]
    repr_type = f.attrs['repr_type']
    out_fmt = f.attrs['output_format']
    mode = f['modes'][0] if 'modes' in f else "Unknown"
    
num_ch = data.shape[0]
fig, axes = plt.subplots(num_ch, 1, figsize=(10, 3*num_ch))

for i in range(num_ch):
    channel_data = data[i]
    
    # Check if it's a magnitude channel (0 or 1)
    if i < 2:
        # Force Log10 for plotting to reveal the hidden linear chirp!
        plot_data = np.log10(np.abs(channel_data) + 1e-30)
        cmap = 'viridis'
        title = f"Channel {i} (Magnitude A/E - Plotted in Log10)"
    else:
        # Phase or Cross-Spectrum channels
        plot_data = channel_data
        cmap = 'twilight' if 'phase' in out_fmt else 'RdBu'
        title = f"Channel {i} (Phase)"
        
    im = axes[i].imshow(plot_data, aspect='auto', origin='lower', cmap=cmap)
    axes[i].set_title(title)
    axes[i].set_ylabel("Freq Bins")
    fig.colorbar(im, ax=axes[i])

axes[-1].set_xlabel("Time Bins")
plt.suptitle(f"Method: {repr_type.upper()} | Output Format: {out_fmt} | Mode: {mode}", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

# Put the exact path to the file you just generated!
h5_path = "/sps/lisaf/mbellotti/github/phd_disentangling/scripts/dataset.h5" 

with h5py.File(h5_path, 'r') as f:
    data = f['data'][0] 
    repr_type = f.attrs['repr_type']
    out_fmt = f.attrs['output_format']
    if 'modes' in f:
        mode = f['modes'][0]
    else:
        mode = "Unknown"
        
num_ch = data.shape[0]
fig, axes = plt.subplots(num_ch, 1, figsize=(12, 3.5 * num_ch), sharex=True)
if num_ch == 1: axes =[axes]

# Set up the proper titles based on your format
if out_fmt == 'real_imag' and num_ch == 4:
    titles =["Mag A", "Mag E", "Cross-Spectrum (Real)", "Cross-Spectrum (Imag)"]
elif out_fmt == 'mag_phase' and num_ch == 3:
    titles =["Mag A", "Mag E", "Phase Diff (A - E)"]
else:
    titles = [f"Channel {i}" for i in range(num_ch)]

for i in range(num_ch):
    channel_data = data[i]
    
    # Magnitudes (Channels 0 and 1)
    if i < 2:
        # Use LogNorm to reveal the linear chirp! (vmin=1% to prevent noise washing it out)
        vmin = np.percentile(channel_data, 1)
        vmax = np.percentile(channel_data, 99.9)
        im = axes[i].imshow(channel_data, aspect='auto', origin='lower', cmap='viridis', 
                            norm=LogNorm(vmin=max(vmin, 1e-10), vmax=max(vmax, 1e-5)))
    
    # Phase or Cross-Spectrum (Channels 2 and 3)
    else:
        # If it's pure phase, it bounds from -pi to pi
        if 'Phase Diff' in titles[i]:
            cmap = 'twilight'
            im = axes[i].imshow(channel_data, aspect='auto', origin='lower', cmap=cmap, vmin=-np.pi, vmax=np.pi)
        # If it's normalized cross spectrum, it bounds from -1 to 1
        else:
            cmap = 'RdBu'
            im = axes[i].imshow(channel_data, aspect='auto', origin='lower', cmap=cmap, vmin=-1.0, vmax=1.0)
            
    axes[i].set_title(titles[i], fontsize=14)
    axes[i].set_ylabel("Frequency Bins", fontsize=12)
    fig.colorbar(im, ax=axes[i], pad=0.01)

axes[-1].set_xlabel("Time Bins", fontsize=12)
plt.suptitle(f"Method: {repr_type.upper()} | Format: {out_fmt} | Mode: {mode}", fontsize=18, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# Put the exact path to the file you just generated!
h5_path = "/sps/lisaf/mbellotti/github/phd_disentangling/scripts/stft_nolog_re_im.h5" 

with h5py.File(h5_path, 'r') as f:
    data = f['data'][0] 
    repr_type = f.attrs['repr_type']
    out_fmt = f.attrs['output_format']
    mode = f['modes'][0] if 'modes' in f else "Unknown"
        
num_ch = data.shape[0]

x = data[0]

print(x.min())
print(x.max())

print(np.percentile(x, [1, 5, 50, 95, 99, 99.9]))

# Create subplots (we add one extra at the bottom for the Ridge Extraction!)
fig, axes = plt.subplots(num_ch + 1, 1, figsize=(12, 3.5 * (num_ch + 1)), sharex=True)

titles = ["Mag A", "Mag E", "Phase Diff (A - E)"] if out_fmt == 'mag_phase' else [f"Channel {i}" for i in range(num_ch)]

for i in range(num_ch):
    channel_data = data[i]
    
    # Magnitudes (Channels 0 and 1)
    if i < 2:
        # DATA IS ALREADY LOG10! Just use vmin/vmax directly.
        vmin = np.percentile(channel_data, 50) # Ignore the bottom 50% background noise
        vmax = np.max(channel_data)            # Top is the chirp
        im = axes[i].imshow(channel_data, aspect='auto', origin='lower', cmap='viridis', vmin=vmin, vmax=vmax)
    
    # Phase or Cross-Spectrum (Channels 2 and 3)
    else:
        if 'Phase' in titles[i] or num_ch == 3:
            cmap = 'twilight'
            im = axes[i].imshow(channel_data, aspect='auto', origin='lower', cmap=cmap, vmin=-np.pi, vmax=np.pi)
        else:
            cmap = 'RdBu'
            im = axes[i].imshow(channel_data, aspect='auto', origin='lower', cmap=cmap, vmin=-1.0, vmax=1.0)
            
    axes[i].set_title(titles[i], fontsize=14)
    axes[i].set_ylabel("Freq Bins", fontsize=12)
    fig.colorbar(im, ax=axes[i], pad=0.01)

# ==========================================
# EXTRACTION OF THE CHIRP RIDGE PHASE
# ==========================================
mag_A = data[0]
phase_diff = data[2]

# For each time bin (column), find the row index of the maximum magnitude (The Ridge)
ridge_indices = np.argmax(mag_A, axis=0)
time_bins = np.arange(mag_A.shape[1])

# Extract the phase difference only along that ridge
extracted_phase = phase_diff[ridge_indices, time_bins]

# We only want to plot the phase where the signal is actually loud (SNR > threshold)
# Otherwise we are just plotting the phase of the loudest noise pixel!
noise_floor = np.percentile(mag_A, 80)
signal_mask = mag_A[ridge_indices, time_bins] > noise_floor

axes[-1].scatter(time_bins[signal_mask], extracted_phase[signal_mask], s=5, c='darkorange', alpha=0.7)
axes[-1].set_title("Extracted Phase Difference Along the Chirp Ridge", fontsize=14)
axes[-1].set_ylabel("Phase Diff (Rads)", fontsize=12)
axes[-1].set_ylim(-np.pi, np.pi)
axes[-1].grid(True, alpha=0.3)

axes[-1].set_xlabel("Time Bins", fontsize=12)
plt.suptitle(f"Method: {repr_type.upper()} | Format: {out_fmt} | Mode: {mode}", fontsize=18, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# Using the file you just generated
h5_path = "/sps/lisaf/mbellotti/github/phd_disentangling/scripts/stft_nolog.h5" 

with h5py.File(h5_path, 'r') as f:
    data = f['data'][0] 
    repr_type = f.attrs['repr_type']
    out_fmt = f.attrs['output_format']

# Get the specific channels based on 'mag_phase_all'
mag_A = data[0]
phase_A = data[2]
phase_diff = data[4]

# Find the chirp ridge (where the signal is loudest)
ridge_indices = np.argmax(mag_A, axis=0)
time_bins = np.arange(mag_A.shape[1])

# Extract the phases only along the ridge
extracted_phase_diff = phase_diff[ridge_indices, time_bins]
extracted_abs_phase = phase_A[ridge_indices, time_bins]

# UNWRAP the absolute phase so it doesn't snap back to -pi!
unwrapped_abs_phase = np.unwrap(extracted_abs_phase)

# Filter out the pure noise bins
noise_floor = np.percentile(mag_A, 80)
signal_mask = mag_A[ridge_indices, time_bins] > noise_floor

# --- PLOTTING ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Phase Difference
axes[0].scatter(time_bins[signal_mask], extracted_phase_diff[signal_mask], s=10, c='royalblue', alpha=0.8)
axes[0].set_title("Phase Difference (A - E)\n(Encodes Sky Location!)", fontsize=14)
axes[0].set_ylabel("Phase (Rads)", fontsize=12)
axes[0].set_xlabel("Time Bins", fontsize=12)
axes[0].set_ylim(-np.pi, np.pi)
axes[0].grid(True, alpha=0.3)

# Plot 2: Unwrapped Absolute Phase
axes[1].scatter(time_bins[signal_mask], unwrapped_abs_phase[signal_mask], s=10, c='darkorange', alpha=0.8)
axes[1].set_title("Unwrapped Absolute Phase (Channel A)\n(Encodes Orbital Evolution)", fontsize=14)
axes[1].set_ylabel("Cumulative Phase (Rads)", fontsize=12)
axes[1].set_xlabel("Time Bins", fontsize=12)
axes[1].grid(True, alpha=0.3)

plt.suptitle("Extracting the Physics from the Chirp Ridge", fontsize=16, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()
```

```python
import glob
import json
import numpy as np
import cupy as cp
import matplotlib.pyplot as plt
from scipy.signal import stft
from scipy.interpolate import interp1d
from matplotlib.colors import LogNorm

from astropy.cosmology import Planck18
from bbhx.waveformbuild import BBHWaveformFD
from bbhx.utils.constants import PC_SI, YRSID_SI
from lisatools.detector import EqualArmlengthOrbits

# ======================================================================
# 1. ALBERTO'S JSON PARSER
# ======================================================================
def extract_alberto_json(filepath):
    with open(filepath, 'r') as f:
        data = json.load(f)
    p = data.get('source_params', data.get('params', data.get('injection', {})))
    
    def get_val(keys, default=0.0):
        for k in keys:
            if k in p: return float(p[k])
        return default

    m1_src, m2_src = get_val(['m1', 'mass1']), get_val(['m2', 'mass2'])
    if m1_src == 0.0:
        mc, q_val = get_val(['mchirp', 'mc']), get_val(['q', 'mass_ratio'])
        m1_src = mc * (1+q_val)**0.2 / q_val**0.6
        m2_src = m1_src * q_val

    z = get_val(['z', 'redshift'], 0.0)
    m1_det, m2_det = m1_src * (1 + z), m2_src * (1 + z)
    m1, m2 = max(m1_det, m2_det), min(m1_det, m2_det)
    
    dist_raw = get_val(['DL_Mpc', 'dL'])
    dist = dist_raw * 1e6 * PC_SI if dist_raw < 1e15 else dist_raw
        
    t_ref = get_val(['Tc', 't_ref'])
    inc = get_val(['inc', 'iota'])
    lam = get_val(['phS', 'lam', 'lambda'])
    beta = np.pi/2.0 - p['thS'] if 'thS' in p else get_val(['beta', 'dec'])
    psi, phi_ref = get_val(['psi', 'pol']), get_val(['ph0', 'phi0'])
    chi1, chi2 = get_val(['chi1', 'spin1'], 0.0), get_val(['chi2', 'spin2'], 0.0)
    
    return [m1, m2, chi1, chi2, dist, phi_ref, inc, lam, beta, psi, t_ref]

# ======================================================================
# 2. LOAD EVENT & GENERATE WAVEFORM
# ======================================================================
repo_path = "/sps/lisaf/mbellotti/github/standard-sirens-MBHBs"
json_file = glob.glob(f"{repo_path}/Pop3/json/*.json")[0]
params = extract_alberto_json(json_file)

print(f"--- Loaded Event ---")
print(f"M1: {params[0]:.2e} M_sun, M2: {params[1]:.2e} M_sun")
print(f"Dist: {params[4]/(PC_SI*1e9):.2f} Gpc, Sky (lam, beta): {params[7]:.2f}, {params[8]:.2f}")

dt = 10.0
T_obs = 7.5 * 24 * 3600
T_gen = 30 * 24 * 3600
N_obs, N_gen = int(T_obs / dt), int(T_gen / dt)

# Force merger to middle of window
params[-1] = T_gen - (T_obs * 0.5) 

# 1. Remove use_gpu=True and force_backend="cuda"
orbits = EqualArmlengthOrbits(use_gpu=False)
wave_gen = BBHWaveformFD(amp_phase_kwargs=dict(run_phenomd=False), response_kwargs=dict(orbits=orbits), force_backend="cpu")

# 2. Change cp (CuPy) to np (Numpy)
freqs_in = np.fft.rfftfreq(N_gen, dt)
wave_out = wave_gen(
    np.array([params[0]]), np.array([params[1]]), np.array([params[2]]), np.array([params[3]]), np.array([params[4]]),
    np.array([params[5]]), np.array([0.0]), np.array([params[6]]), np.array([params[7]]), np.array([params[8]]), 
    np.array([params[9]]), np.array([params[-1]]),
    freqs=freqs_in, modes=[(2,2), (2,1), (3,3), (3,2), (4,4), (4,3)], direct=False, fill=True, length=N_gen, t_obs_start=0.0, t_obs_end=T_gen/YRSID_SI
)

# 3. Use numpy directly (remove the .get())
hA = np.fft.irfft(wave_out[0, 0, :], n=N_gen)[-N_obs:] * 1e22
hE = np.fft.irfft(wave_out[0, 1, :], n=N_gen)[-N_obs:] * 1e22

# ======================================================================
# 3. STFT & LOG-INTERPOLATION
# ======================================================================
f_lin, t_stft, Z_A = stft(hA, fs=1/dt, nperseg=1024, noverlap=900)
_, _, Z_E = stft(hE, fs=1/dt, nperseg=1024, noverlap=900)

# Remove DC component for log scale
f_lin, Z_A, Z_E = f_lin[1:], Z_A[1:, :], Z_E[1:, :]

# Extract Magnitude and Phase
mag_A, mag_E = np.abs(Z_A), np.abs(Z_E)

# CRITICAL: Unwrap phase along the Frequency axis (axis=0) so interpolation is safe!
phase_A = np.unwrap(np.angle(Z_A), axis=0)
phase_E = np.unwrap(np.angle(Z_E), axis=0)

# Interpolate to Log-Frequency
n_freq = 128
f_log = np.logspace(np.log10(f_lin[0]), np.log10(f_lin[-1]), num=n_freq)

interp_mag_A = interp1d(f_lin, mag_A, axis=0, kind='linear', fill_value="extrapolate")
interp_mag_E = interp1d(f_lin, mag_E, axis=0, kind='linear', fill_value="extrapolate")
interp_phase_A = interp1d(f_lin, phase_A, axis=0, kind='linear', fill_value="extrapolate")
interp_phase_E = interp1d(f_lin, phase_E, axis=0, kind='linear', fill_value="extrapolate")

mag_A_log = interp_mag_A(f_log)
mag_E_log = interp_mag_E(f_log)
phase_A_log = interp_phase_A(f_log)
phase_E_log = interp_phase_E(f_log)

# ======================================================================
# 4. DEEP LEARNING FEATURES
# ======================================================================
log_mag_A = np.log10(mag_A_log + 1e-30)

# Phase Difference tells us the geometry!
phase_diff = phase_A_log - phase_E_log

# Normalized Cross Spectrum Real and Imaginary (Cos and Sin)
cross_re = np.cos(phase_diff)
cross_im = np.sin(phase_diff)

# For visualization: Unwrap Phase A along the Time axis (axis=1) to show it growing
absolute_phase_A_growing = np.unwrap(phase_A_log, axis=1)

# ======================================================================
# 5. PLOTTING
# ======================================================================
fig, axes = plt.subplots(5, 1, figsize=(10, 16), sharex=True)

# 1. Magnitude
im0 = axes[0].pcolormesh(t_stft/3600, f_log, log_mag_A, cmap='viridis', shading='auto')
axes[0].set(title="1. Log-Magnitude (Channel A)", ylabel="Freq [Hz]", yscale='log')
fig.colorbar(im0, ax=axes[0])

# 2. Absolute Phase A (Growing over time)
im1 = axes[1].pcolormesh(t_stft/3600, f_log, absolute_phase_A_growing, cmap='magma', shading='auto')
axes[1].set(title="2. Absolute Phase A (Unwrapped over time - See it grow!)", ylabel="Freq [Hz]", yscale='log')
fig.colorbar(im1, ax=axes[1])

# 3. Phase Difference A - E (For 'mag_phase' format)
im2 = axes[2].pcolormesh(t_stft/3600, f_log, phase_diff, cmap='twilight', shading='auto', vmin=-np.pi, vmax=np.pi)
axes[2].set(title="3. Phase Difference (A - E) [Used for 'mag_phase']", ylabel="Freq [Hz]", yscale='log')
fig.colorbar(im2, ax=axes[2])

# 4. Cross Spectrum Real (For 'real_imag' format)
im3 = axes[3].pcolormesh(t_stft/3600, f_log, cross_re, cmap='RdBu', shading='auto', vmin=-1, vmax=1)
axes[3].set(title="4. Cross-Spectrum Real (Cos \u0394\u03A6) [Used for 'real_imag']", ylabel="Freq [Hz]", yscale='log')
fig.colorbar(im3, ax=axes[3])

# 5. Cross Spectrum Imag (For 'real_imag' format)
im4 = axes[4].pcolormesh(t_stft/3600, f_log, cross_im, cmap='RdBu', shading='auto', vmin=-1, vmax=1)
axes[4].set(title="5. Cross-Spectrum Imag (Sin \u0394\u03A6) [Used for 'real_imag']", xlabel="Time [hours]", ylabel="Freq [Hz]", yscale='log')
fig.colorbar(im4, ax=axes[4])

plt.tight_layout()
plt.show()
```

```python
import glob
import json
import numpy as np
import torch
import cupy as cp
import matplotlib.pyplot as plt
from scipy.signal import stft
from matplotlib.colors import LogNorm

from astropy.cosmology import Planck18
from bbhx.waveformbuild import BBHWaveformFD
from bbhx.utils.constants import PC_SI, YRSID_SI
from lisatools.detector import EqualArmlengthOrbits
from nnAudio.features.cqt import CQT1992v2

# ======================================================================
# 1. ALBERTO'S JSON PARSER
# ======================================================================
def extract_alberto_json(filepath):
    with open(filepath, 'r') as f:
        data = json.load(f)
    p = data.get('source_params', data.get('params', data.get('injection', {})))
    
    def get_val(keys, default=0.0):
        for k in keys:
            if k in p: return float(p[k])
        return default

    m1_src = get_val(['m1', 'mass1', 'M1'])
    m2_src = get_val(['m2', 'mass2', 'M2'])
    if m1_src == 0.0 and m2_src == 0.0:
        mc, q_val = get_val(['mchirp', 'mc']), get_val(['q', 'mass_ratio'])
        m1_src = mc * (1+q_val)**0.2 / q_val**0.6
        m2_src = m1_src * q_val

    z = get_val(['z', 'redshift'], 0.0)
    m1_det, m2_det = m1_src * (1 + z), m2_src * (1 + z)
    m1, m2 = max(m1_det, m2_det), min(m1_det, m2_det)
    q = m2 / m1 if m1 > 0 else 1.0
    
    dist_raw = get_val(['DL_Mpc', 'dL', 'distance'])
    dist = dist_raw * 1e6 * PC_SI if dist_raw < 1e15 else dist_raw
        
    t_ref = get_val(['Tc', 'tC', 'tc', 't_ref', 'time'])
    inc = get_val(['inc', 'iota'])
    lam = get_val(['phS', 'lam', 'lambda', 'ra'])
    beta = np.pi/2.0 - p['thS'] if 'thS' in p else get_val(['beta', 'dec'])
    psi = get_val(['psi', 'pol'])
    phi_ref = get_val(['ph0', 'phi0', 'phi_ref'])
    chi1, chi2 = get_val(['chi1', 'spin1'], 0.0), get_val(['chi2', 'spin2'], 0.0)
    
    return[m1, m2, chi1, chi2, dist, phi_ref, inc, lam, beta, psi, t_ref]

# ======================================================================
# 2. LOAD EVENT & GENERATE WAVEFORM
# ======================================================================
#repo_path = "../../standard-sirens-MBHBs"
repo_path = "/sps/lisaf/mbellotti/github/standard-sirens-MBHBs"
json_file = glob.glob(f"{repo_path}/Pop3/json/*.json")[0]
params = extract_alberto_json(json_file)

print(f"--- Loaded Event ---")
print(f"M1: {params[0]:.2e} M_sun, M2: {params[1]:.2e} M_sun")
print(f"Dist: {params[4]/(PC_SI*1e9):.2f} Gpc, Sky (lam, beta): {params[7]:.2f}, {params[8]:.2f}")

dt = 10.0
T_obs = 7.5 * 24 * 3600
T_gen = 30 * 24 * 3600
N_obs, N_gen = int(T_obs / dt), int(T_gen / dt)

# Force the merger to happen in the middle of our 7.5 day window!
params[-1] = T_gen - (T_obs * 0.5) 

# orbits = EqualArmlengthOrbits(use_gpu=True, force_backend="cuda")
# wave_gen = BBHWaveformFD(amp_phase_kwargs=dict(run_phenomd=False), response_kwargs=dict(orbits=orbits), force_backend="cuda")
orbits = EqualArmlengthOrbits(use_gpu=False) # Removed force_backend="cuda"
wave_gen = BBHWaveformFD(amp_phase_kwargs=dict(run_phenomd=False), response_kwargs=dict(orbits=orbits)) 

freqs_in = cp.asarray(np.fft.rfftfreq(N_gen, dt))
wave_out = wave_gen(
    cp.array([params[0]]), cp.array([params[1]]), cp.array([params[2]]), cp.array([params[3]]), cp.array([params[4]]),
    cp.array([params[5]]), cp.array([0.0]), cp.array([params[6]]), cp.array([params[7]]), cp.array([params[8]]), 
    cp.array([params[9]]), cp.array([params[-1]]),
    freqs=freqs_in, modes=[(2,2), (2,1), (3,3), (3,2), (4,4), (4,3)], direct=False, fill=True, length=N_gen, t_obs_start=0.0, t_obs_end=T_gen/YRSID_SI
)

hA = cp.fft.irfft(wave_out[0, 0, :], n=N_gen)[-N_obs:].get()
hE = cp.fft.irfft(wave_out[0, 1, :], n=N_gen)[-N_obs:].get()

print(f"Waveform Max Amplitude: {np.max(np.abs(hA)):.2e}")
if np.isnan(hA).any(): print("WARNING: NaNs detected in generated waveform!")

# ======================================================================
# 3. TIME-FREQUENCY REPRESENTATIONS
# ======================================================================
# --- STFT ---
f_stft, t_stft, Z_stft_A = stft(hA, fs=1/dt, nperseg=1024, noverlap=900)
_, _, Z_stft_E = stft(hE, fs=1/dt, nperseg=1024, noverlap=900)
mag_stft = np.abs(Z_stft_A) + 1e-30
cross_stft = Z_stft_A * np.conj(Z_stft_E)
phase_diff_stft = np.angle(cross_stft)

# --- CQT (nnAudio) ---
cqt_layer = CQT1992v2(sr=1/dt, fmin=1e-4, fmax=5e-2, hop_length=128, bins_per_octave=16, output_format="Complex", trainable=False).to("cuda")
mix_tensor = torch.tensor(np.stack([hA, hE]), dtype=torch.float32, device="cuda")
cqt_raw = cqt_layer(mix_tensor)
cqt_complex = torch.complex(cqt_raw[..., 0], cqt_raw[..., 1]).cpu().numpy()

CQT_A, CQT_E = cqt_complex[0], cqt_complex[1]
mag_cqt = np.abs(CQT_A) + 1e-30
cross_cqt = CQT_A * np.conj(CQT_E)
phase_diff_cqt = np.angle(cross_cqt)

# ======================================================================
# 4. PLOTTING
# ======================================================================
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Secure VMIN and VMAX to completely prevent "all white"
stft_vmin, stft_vmax = np.percentile(mag_stft, 1), np.percentile(mag_stft, 99.9)
cqt_vmin, cqt_vmax = np.percentile(mag_cqt, 1), np.percentile(mag_cqt, 99.9)

im0 = axes[0, 0].pcolormesh(t_stft/3600, f_stft, mag_stft, norm=LogNorm(vmin=stft_vmin, vmax=stft_vmax), cmap='viridis', shading='auto')
axes[0, 0].set(title="STFT Magnitude (Channel A)", ylabel="Freq [Hz]", ylim=(1e-4, 1e-2), yscale='log')
fig.colorbar(im0, ax=axes[0,0])

im1 = axes[0, 1].pcolormesh(t_stft/3600, f_stft, phase_diff_stft, cmap='twilight', shading='auto', vmin=-np.pi, vmax=np.pi)
axes[0, 1].set(title="STFT Phase Diff (A - E)", ylim=(1e-4, 1e-2), yscale='log')
fig.colorbar(im1, ax=axes[0,1])

t_cqt_hours = np.arange(mag_cqt.shape[1]) * (128 * dt) / 3600.0
f_cqt = cqt_layer.frequencies

im2 = axes[1, 0].pcolormesh(t_cqt_hours, f_cqt, mag_cqt, norm=LogNorm(vmin=cqt_vmin, vmax=cqt_vmax), cmap='viridis', shading='auto')
axes[1, 0].set(title="CQT Magnitude (Channel A)", xlabel="Time[hours]", ylabel="Freq [Hz]")
fig.colorbar(im2, ax=axes[1,0])

im3 = axes[1, 1].pcolormesh(t_cqt_hours, f_cqt, phase_diff_cqt, cmap='twilight', shading='auto', vmin=-np.pi, vmax=np.pi)
axes[1, 1].set(title="CQT Phase Diff (A - E)", xlabel="Time [hours]")
fig.colorbar(im3, ax=axes[1,1])

plt.tight_layout()
plt.savefig("mbhb_time_freq_representations.png", dpi=300)
plt.show()
```
